```python
from langchain_core.documents import Document
```

```python
docs = [
    Document(page_content="Langchain makes it way to work with LLMs"),
    Document(page_content="Langchain is used to build LLM based application"),
    Document(page_content="Chroma is used to store and search document embedding"),
    Document(page_content="Langchain support Chroma, FAISS, Pincore, and many more"),
]
```

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

embedding_model = HuggingFaceEmbeddings()

vectorstore = FAISS.from_documents(
    documents=docs,
    embedding=embedding_model
)
```

```python
from re import search
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 3 , "lambda_mult": 1}
)
```

```python
from redis.commands.search import query
query = "What is langgraph?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content) 
```
