```python
import wikipedia
from langchain_community.retrievers import WikipediaRetriever

# Set custom User-Agent and rate limiting to avoid Wikimedia rate-limiting / blocking
wikipedia.set_user_agent("LangChainApp/1.0 (contact@example.com)")
wikipedia.set_rate_limiting(True)
```

```python
# Initialize the retrievers (optional: set the language and top_k)
retriever = WikipediaRetriever(top_k_results=2, lang="en")
```

```python
# Define your query
query = "the geopolitical history of india and pakistan from the perspective of a chinese"

# Get relevant wikipedia documents
docs = retriever.invoke(query)
```

```python
# print retrieved content
for i, doc in enumerate(docs):
    print(f"\n--- Result {i+1} ---")
    print(f"Content:\n{doc.page_content}...") # truncate for display
```
