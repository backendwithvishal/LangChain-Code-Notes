```python
# All the import that need
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEndpoint
from langchain_classic.retrievers import MultiQueryRetriever
```

```python
# Relevant health & wellness documents
all_docs = [
    Document(
        page_content="LangChain is a framework for building applications powered by large language models.",
        metadata={"source": "langchain_intro"}
    ),
    Document(
        page_content="Retrieval Augmented Generation (RAG) retrieves relevant documents and provides them to an LLM to generate better answers.",
        metadata={"source": "rag_intro"}
    ),
    Document(
        page_content="MultiQueryRetriever generates multiple versions of a user's query to retrieve relevant documents from different perspectives.",
        metadata={"source": "multi_query"}
    ),
    Document(
        page_content="Vector databases store document embeddings and allow similarity-based searches to find relevant information.",
        metadata={"source": "vector_database"}
    ),
    Document(
        page_content="Hugging Face provides open-source models and tools that can be used for text embeddings and natural language processing.",
        metadata={"source": "huggingface"}
    ),
]
```

```python
# Initialize HuggingFaceAPIKey Embedding
embedding_model = HuggingFaceEmbeddings()

# Create FAISS vector store
vectorstore = FAISS.from_documents(documents=all_docs, embedding=embedding_model)
```

```python
# Create retrievers
similarity_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)
```

```python
multiquery_retriever = MultiQueryRetriever.from_llm(
    retriever=vectorstore.as_retriever(
        search_kwargs={"k": 5}
    ),
    llm=HuggingFaceEndpoint(
        repo_id="Qwen/Qwen2.5-7B-Instruct"
    )
)
```

```python
# Query
query = "How to imporve RAG in security side"
```

```python
# Retriever results
similarity_results = similarity_retriever.invoke(query)
multiquery_results = multiquery_retriever.invoke(query)
```
